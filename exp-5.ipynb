```python

import cv2
import matplotlib.pyplot as plt
import numpy as np

# Read the input image
image1 = cv2.imread("taj.jpg")

# Check whether image is loaded
if image1 is None:
    print("Error: Image not found!")
else:

    # Convert BGR to RGB
    image2 = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)

    # 1. Average Filter
    kernel = np.ones((11, 11), np.float32) / 121
    average = cv2.filter2D(image2, -1, kernel)

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(image2)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(average)
    plt.title("Average Filter Image")
    plt.axis("off")
    plt.show()

    # 2. Weighted Average Filter
    kernel1 = np.array([[1, 2, 1],
                        [2, 4, 2],
                        [1, 2, 1]]) / 16

    weighted = cv2.filter2D(image2, -1, kernel1)

    plt.imshow(weighted)
    plt.title("Weighted Average Filter Image")
    plt.axis("off")
    plt.show()

    # 3. Gaussian Blur
    gaussian_blur = cv2.GaussianBlur(image2, (33, 33), 0)

    plt.imshow(gaussian_blur)
    plt.title("Gaussian Blur")
    plt.axis("off")
    plt.show()

    # 4. Median Blur
    median = cv2.medianBlur(image2, 13)

    plt.imshow(median)
    plt.title("Median Blur")
    plt.axis("off")
    plt.show()

    # 5. Laplacian Kernel
    kernel2 = np.array([[-1, -1, -1],
                        [ 2, -2,  1],
                        [ 2,  1, -1]], dtype=np.float32)

    laplacian_kernel = cv2.filter2D(
        image2, cv2.CV_32F, kernel2
    )

    # Convert result for display
    laplacian_kernel = cv2.convertScaleAbs(laplacian_kernel)

    plt.imshow(laplacian_kernel)
    plt.title("Laplacian Kernel")
    plt.axis("off")
    plt.show()

    # 6. Laplacian Operator
    laplacian = cv2.Laplacian(image2, cv2.CV_64F)
    laplacian = cv2.convertScaleAbs(laplacian)

    plt.imshow(laplacian)
    plt.title("Laplacian Operator")
    plt.axis("off")
    plt.show()
```
